### **Create a Count plot**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Calculate Ship Mode distribution
ship_distribution = dataset['Ship Mode'].value_counts()

# Create Area Chart
plt.figure(figsize=(10, 6))

plt.fill_between(
    ship_distribution.index,
    ship_distribution.values,
    alpha=0.4,
    color='skyblue'
)

plt.plot(
    ship_distribution.index,
    ship_distribution.values,
    marker='o',
    linewidth=2,
    color='blue'
)

# Add chart title and labels
plt.title('Area Chart - Ship Mode Distribution', fontsize=15)
plt.xlabel('Ship Mode', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# Display values
for i, value in enumerate(ship_distribution.values):
    plt.text(i, value, str(value), ha='center', va='bottom')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Group data by Ship Mode
bubble_data = dataset.groupby('Ship Mode').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity ordered new': 'sum'
}).reset_index()

# Create Bubble Chart
plt.figure(figsize=(12, 7))

plt.scatter(
    bubble_data['Sales'],
    bubble_data['Profit'],
    s=bubble_data['Quantity ordered new'] * 2,
    alpha=0.6,
    c=bubble_data['Quantity ordered new'],
    cmap='viridis',
    edgecolors='black'
)

# Add Ship Mode labels
for i, row in bubble_data.iterrows():
    plt.annotate(
        row['Ship Mode'],
        (row['Sales'], row['Profit']),
        xytext=(5, 5),
        textcoords='offset points',
        fontsize=10
    )

# Chart formatting
plt.title('Bubble Chart: Sales vs Profit by Ship Mode',
          fontsize=15)

plt.xlabel('Total Sales', fontsize=12)
plt.ylabel('Total Profit', fontsize=12)

plt.colorbar(label='Total Quantity Ordered')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total profit by Product Category
category_profit = dataset.groupby('Product Category')['Profit'].sum().sort_values(ascending=False)

# Create bar chart
plt.figure(figsize=(9, 5))

sns.barplot(
    x=category_profit.index,
    y=category_profit.values,
    errorbar=None
)

plt.title('Product Category-wise Total Profit', fontsize=14)
plt.xlabel('Product Category')
plt.ylabel('Total Profit')

# Display values
for i, value in enumerate(category_profit.values):
    plt.text(i, value, f'{value:,.0f}', ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

```python
# Calculate total sales by City
city_sales = dataset.groupby('City')['Sales'].sum().sort_values(ascending=False)

print(city_sales.head(10))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get top 10 cities
top_10_cities = city_sales.head(10)

# Create bar chart
plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_cities.values,
    y=top_10_cities.index
)

plt.title('Top 10 Cities by Total Sales', fontsize=15)
plt.xlabel('Total Sales')
plt.ylabel('City')

# Display sales values
for i, value in enumerate(top_10_cities.values):
    plt.text(value, i, f' {value:,.0f}', va='center')

plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total sales by Region
region_sales = dataset.groupby('Region')['Sales'].sum().sort_values(ascending=False)

print("Region-wise Total Sales:")
print(region_sales)

# Find highest sales region
highest_region = region_sales.idxmax()
highest_sales = region_sales.max()

print("\nHighest Sales Region:", highest_region)
print("Total Sales:", highest_sales)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    x=region_sales.index,
    y=region_sales.values
)

plt.title('Region-wise Total Sales', fontsize=15)
plt.xlabel('Region')
plt.ylabel('Total Sales')

# Display sales values
for i, value in enumerate(region_sales.values):
    plt.text(i, value, f'{value:,.0f}', ha='center', va='bottom')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
import pandas as pd

# Filter loss-making transactions
loss_data = dataset[dataset['Profit'] < 0]

# Calculate total loss
total_loss = abs(loss_data['Profit'].sum())

print("Total Loss:", total_loss)
print("Total Loss-making Transactions:", len(loss_data))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Filter loss-making transactions
loss_data = dataset[dataset['Profit'] < 0]

# Calculate total loss by Region
region_loss = loss_data.groupby('Region')['Profit'].sum().abs()

# Sort losses
region_loss = region_loss.sort_values(ascending=False)

# Create Pie Chart
plt.figure(figsize=(9, 7))

plt.pie(
    region_loss.values,
    labels=region_loss.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=[0.05] * len(region_loss),
    shadow=False,
    textprops={'fontsize': 11}
)

plt.title(
    'Region-wise Total Loss Distribution',
    fontsize=15
)

plt.legend(
    region_loss.index,
    title='Regions',
    loc='best'
)

plt.tight_layout()
plt.show()

# Display region-wise loss
print("Region-wise Total Loss:")
print(region_loss)
```

```python
category_loss = loss_data.groupby('Product Category')['Profit'].sum().abs().sort_values(ascending=False)

plt.figure(figsize=(9, 6))

plt.bar(
    category_loss.index,
    category_loss.values
)

plt.title('Product Category-wise Loss')
plt.xlabel('Product Category')
plt.ylabel('Total Loss')

plt.tight_layout()
plt.show()

print(category_loss)
```

```python
city_loss = loss_data.groupby('City')['Profit'].sum().sort_values()

top_10_loss_cities = city_loss.head(10).abs()

plt.figure(figsize=(12, 6))

sns.barplot(
    x=top_10_loss_cities.values,
    y=top_10_loss_cities.index,
    errorbar=None
)

plt.title('Top 10 Cities with Highest Loss')
plt.xlabel('Total Loss')
plt.ylabel('City')

plt.tight_layout()
plt.show()

print(top_10_loss_cities)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total profit by Region
region_profit = dataset.groupby('Region')['Profit'].sum().sort_values(ascending=False)

# Create Bar Chart
plt.figure(figsize=(10, 6))

sns.barplot(
    x=region_profit.index,
    y=region_profit.values,
    errorbar=None
)

# Add data labels
for i, value in enumerate(region_profit.values):
    plt.text(
        i,
        value,
        f'{value:,.0f}',
        ha='center',
        va='bottom' if value >= 0 else 'top'
    )

# Chart formatting
plt.title('Region-wise Profit Distribution', fontsize=15)
plt.xlabel('Region', fontsize=12)
plt.ylabel('Total Profit', fontsize=12)

plt.axhline(y=0, color='black', linewidth=0.8)
plt.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

# Display profit data
print("Region-wise Total Profit:")
print(region_profit)
```
