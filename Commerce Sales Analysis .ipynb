```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Calculate category counts
category_count = dataset["Product Category"].value_counts()

# Convert to 2D DataFrame
heatmap_data = category_count.to_frame(name="Order Count")

# Create heatmap
plt.figure(figsize=(7, 4))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt="d",
    cmap="YlGnBu",
    linewidths=1,
    cbar=True
)

plt.title("Product Category Distribution Heatmap")
plt.xlabel("Order Count")
plt.ylabel("Product Category")

plt.tight_layout()
plt.show()
```

### **Create a Count plot**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.countplot(data=dataset, x='Ship Mode')

plt.title('Distribution of Ship Modes')
plt.xlabel('Ship Mode')
plt.ylabel('Count')

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Calculate Ship Mode distribution
ship_distribution = dataset['Ship Mode'].value_counts()

# Create Area Chart
plt.figure(figsize=(10, 6))

plt.fill_between(
    ship_distribution.index,
    ship_distribution.values,
    alpha=0.4,
    color='skyblue'
)

plt.plot(
    ship_distribution.index,
    ship_distribution.values,
    marker='o',
    linewidth=2,
    color='blue'
)

# Add chart title and labels
plt.title('Area Chart - Ship Mode Distribution', fontsize=15)
plt.xlabel('Ship Mode', fontsize=12)
plt.ylabel('Number of Orders', fontsize=12)

# Display values
for i, value in enumerate(ship_distribution.values):
    plt.text(i, value, str(value), ha='center', va='bottom')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Group data by Ship Mode
bubble_data = dataset.groupby('Ship Mode').agg({
    'Sales': 'sum',
    'Profit': 'sum',
    'Quantity ordered new': 'sum'
}).reset_index()

# Create Bubble Chart
plt.figure(figsize=(12, 7))

plt.scatter(
    bubble_data['Sales'],
    bubble_data['Profit'],
    s=bubble_data['Quantity ordered new'] * 2,
    alpha=0.6,
    c=bubble_data['Quantity ordered new'],
    cmap='viridis',
    edgecolors='black'
)

# Add Ship Mode labels
for i, row in bubble_data.iterrows():
    plt.annotate(
        row['Ship Mode'],
        (row['Sales'], row['Profit']),
        xytext=(5, 5),
        textcoords='offset points',
        fontsize=10
    )

# Chart formatting
plt.title('Bubble Chart: Sales vs Profit by Ship Mode',
          fontsize=15)

plt.xlabel('Total Sales', fontsize=12)
plt.ylabel('Total Profit', fontsize=12)

plt.colorbar(label='Total Quantity Ordered')

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
